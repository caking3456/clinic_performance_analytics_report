```python
import pandas as pd
from faker import Faker
import random
from datetime import datetime, timedelta

# Initialize Faker and set seed for reproducibility
fake = Faker()
random.seed(42)
Faker.seed(42)

# Define specialties and appointment reasons
SPECIALTIES = [
    "Cardiology", "Pediatrics", "Oncology", "Surgery", "Emergency",
    "Radiology", "Neurology", "Orthopedics", "Dermatology", "Psychiatry"
]

APPOINTMENT_REASONS = [
    "Annual Checkup", "Follow-up", "Lab Results", "Vaccination",
    "Consultation", "Emergency Visit", "Physical Therapy"
]

# Generate Patients Table (20 rows)
def generate_patients(num_patients=20):
    patients = []
    for patient_id in range(1, num_patients + 1):
        first_name = fake.first_name()
        last_name = fake.last_name()
        dob = fake.date_of_birth(minimum_age=18, maximum_age=90)
        gender = random.choice(["Male", "Female", "Other"])
        phone = fake.phone_number()
        email = fake.email()
        
        patients.append({
            "PatientID": patient_id,
            "FirstName": first_name,
            "LastName": last_name,
            "DOB": dob.strftime("%Y-%m-%d"),
            "Gender": gender,
            "Phone": phone,
            "Email": email
        })
    return pd.DataFrame(patients)

# Generate Providers Table (10 rows)
def generate_providers(num_providers=10):
    providers = []
    for provider_id in range(1, num_providers + 1):
        first_name = fake.first_name()
        last_name = fake.last_name()
        specialty = random.choice(SPECIALTIES)
        
        providers.append({
            "ProviderID": provider_id,
            "FirstName": first_name,
            "LastName": last_name,
            "Specialty": specialty
        })
    return pd.DataFrame(providers)

# Generate Appointments Table (30 rows)
def generate_appointments(num_appointments=30, num_patients=20, num_providers=10):
    appointments = []
    start_date = datetime(2025, 5, 1)
    
    for appointment_id in range(1, num_appointments + 1):
        patient_id = random.randint(1, num_patients)
        provider_id = random.randint(1, num_providers)
        days_offset = random.randint(0, 29)
        appointment_date = (start_date + timedelta(days=days_offset)).strftime("%Y-%m-%d")
        reason = random.choice(APPOINTMENT_REASONS)
        
        appointments.append({
            "AppointmentID": appointment_id,
            "PatientID": patient_id,
            "ProviderID": provider_id,
            "AppointmentDate": appointment_date,
            "Reason": reason
        })
    return pd.DataFrame(appointments)

# Generate all tables
patients_df = generate_patients()
providers_df = generate_providers()
appointments_df = generate_appointments()

# Export to CSV
patients_df.to_csv("patients.csv", index=False)
providers_df.to_csv("providers.csv", index=False)
appointments_df.to_csv("appointments.csv", index=False)

print("Datasets generated: patients.csv, providers.csv, appointments.csv")
```
